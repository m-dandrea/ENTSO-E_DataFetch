```python
from entsoe import EntsoePandasClient
import pandas as pd
from itertools import permutations
from entsoe.mappings import NEIGHBOURS
import time
import numpy as np
```

Configuration

```python
with open("API_KEY.txt", "r") as f:
    API_KEY = f.read().strip()
client = EntsoePandasClient(api_key=API_KEY,
                            timeout=5, retry_count=3, retry_delay=5)
```

```python
# Define the country codes
country_codes = [
                'AL', 'AT', 'BA', 'BE', 'BG', 'CH', 'CZ', 'DE', 'DK', 'EE', 'ES', 'FI', 
                'FR', 'HR', 'HU', 'IE', 'LT', 'MK', 'NO', 'RO', 'ME', 'PL', 'GR', 'IT', 
                'NL', 'CY', 'SI', 'RS', 'SK', 'LU', 'SE', 'LV', 'PT', 'UK', 'MT'
                ]

#type_marketagreement_type = 'A01'
#contract_marketagreement_type = "A01"
#process_type = 'A51'

# Specify the date range, used in the output filename
year= 2024
start = pd.Timestamp(f'{year}-01-01', tz='Europe/Brussels')
end = pd.Timestamp(f'{year}-12-31', tz='Europe/Brussels')
```

Net Transfer capacity - Day Ahead (MW)

```python
# Initialize a list to store summary rows
summary_rows = []

for country_code_from, country_neighbors in NEIGHBOURS.items():
    for country_code_to in country_neighbors:
        print(f"{country_code_from} -> {country_code_to}")

        try:
            param_NTC = client.query_net_transfer_capacity_dayahead(
                country_code_from, country_code_to, start=start, end=end
            )

            mean_ntc = param_NTC.mean()
            max_ntc = param_NTC.max()

            summary_rows.append({
                "country_code_from": country_code_from,
                "country_code_to": country_code_to,
                "mean_NTC_MW": mean_ntc,
                "max_NTC_MW": max_ntc,
            })

        except Exception as e:
            print(f"Error for {country_code_from} -> {country_code_to}: {str(e)}")
            continue

        time.sleep(1)

# Create summary DataFrame
NTC_summary_df = pd.DataFrame(summary_rows)

# Save to Excel
#NTC_summary_df.to_excel("entsoe_ntc_summary.xlsx", index=False)
NTC_summary_df.to_csv(r'saved_data/' + f"NTC_dayahead_df_{year}.csv", sep='\t', encoding='utf-8')
print(f"Saved to NTC_dayahead_df_{year}.csv")
```

Cross-border Flow (MW)

```python
# Initialize a list to store summary rows
summary_rows = []

for country_code_from, country_neighbors in NEIGHBOURS.items():
    for country_code_to in country_neighbors:
        print(f"{country_code_from} -> {country_code_to}")

        try:
            param_cb = client.query_crossborder_flows(
                country_code_from, country_code_to, start=start, end=end
            )

            mean_cb = param_cb.mean()
            max_cb = param_cb.max()

            summary_rows.append({
                "country_code_from": country_code_from,
                "country_code_to": country_code_to,
                "mean_cross_border_MW": mean_cb,
                "max_cross_border_MW": max_cb,
            })

        except Exception as e:
            print(f"Error for {country_code_from} -> {country_code_to}: {str(e)}")
            continue

        time.sleep(1)

# Create summary DataFrame
cb_summary_df = pd.DataFrame(summary_rows)

# Save to Excel
#NTC_summary_df.to_excel("entsoe_ntc_summary.xlsx", index=False)
cb_summary_df.to_csv(r'saved_data/' + f"cross_border_{year}.csv", sep='\t', encoding='utf-8')

```

```python
import requests
from io import BytesIO
import os

# Set the output directory as an external variable
output_dir = "saved_data/before_2015"
os.makedirs(output_dir, exist_ok=True)

# Set this variable to True to save as CSV, or False to save as XLSX
save_as_csv = True

for r in country_codes:
    url = f"https://transparency.entsoe.eu/content/static_content/getLegacyDocument?path=/Legacy%20data/TRANSMISSION/CROSS-BORDER-PHYSICAL-FLOW/2011/NET_CROSS_BORDER_PHYSICAL_FLOWS_201101010000-201201010000_{r}.xlsx"
    response = requests.get(url)

    if response.status_code == 200:
        try:
            if save_as_csv:
                df = pd.read_excel(BytesIO(response.content), engine='openpyxl')
                csv_file = os.path.join(output_dir, f"cross_border_schedule_2011_{r}.csv")
                df.to_csv(csv_file, index=False)
                print(f"File converted to csv and saved successfully for {r}!")
            else:
                xlsx_file = os.path.join(output_dir, f"cross_border_schedule_2011_{r}.xlsx")
                with open(xlsx_file, "wb") as f:
                    f.write(response.content)
                print(f"File saved as xlsx for {r}!")
        except Exception as e:
            print(f"Failed to process file for {r}: {str(e)}")
    else:
        print(f"Failed to download file for {r}: {response.status_code}")
```

Process the downloaded data and output an aggregated CSV file.

```python

import pandas as pd
from pathlib import Path

# --- Settings ---
FOLDER = Path("saved_data/before_2015")
OUTDIR = FOLDER / "processed"
OUTDIR.mkdir(parents=True, exist_ok=True)

def process_file(fp: Path) -> pd.DataFrame:
    # Read CSV, skip the first 3 rows (header/banner)
    df = pd.read_csv(fp, header=None, encoding="cp1252", dtype=str, skiprows=3)
    if df.empty:
        raise ValueError("No usable data rows after skipping first 3 rows")

    # Transpose & promote first row to header
    df = df.T
    df.columns = df.iloc[0]
    df = df.iloc[1:].reset_index(drop=True)

    # The first column is always the Time Interval -> rename it to "Lines"
    df = df.rename(columns={df.columns[0]: "Lines"})

    # Unpivot (melt) all other columns
    long = df.melt(id_vars=["Lines"], var_name="Attribute", value_name="Value")

    # Drop empty Lines
    long = long[long["Lines"].notna() & (long["Lines"].astype(str).str.strip() != "")].copy()

    # Remove textual invalids (N/E, N/A)
    val_str = long["Value"].astype(str).str.strip().str.lower()
    long = long[~val_str.isin({"n/e", "n/a"})].copy()

    # Convert to numeric and drop NaNs and zeros
    long["Value"] = pd.to_numeric(long["Value"], errors="coerce")
    long = long.dropna(subset=["Value"])
    long = long[long["Value"] != 0]

    if long.empty:
        raise ValueError("All values filtered out (only N/E, N/A, zeros, or non-numeric).")

    # Aggregate (Mean, Max) per line and sort by Max desc
    agg = (
        long.groupby("Lines", as_index=False)
            .agg(Mean=("Value", "mean"), Max=("Value", "max"))
            .sort_values("Max", ascending=False)
            .reset_index(drop=True)
    )

    # Split "Lines" like "DK1 > DE" into from/to (keep raw tokens)
    split = agg["Lines"].astype(str).str.split(">", n=1, expand=True)
    if split.shape[1] < 2:
        # If malformed, add empty right side
        split = pd.concat([split, pd.Series([""] * len(split), index=split.index)], axis=1)

    agg["country_from"] = split[0].str.replace("\u00A0", " ", regex=False).str.strip()
    agg["country_to"]   = split[1].str.replace("\u00A0", " ", regex=False).str.strip()

    # Add source file and return tidy columns
    agg["source_file"] = fp.name
    return agg[["country_from", "country_to", "Mean", "Max", "Lines", "source_file"]]

def main():
    files = sorted(FOLDER.glob("*.csv"))
    if not files:
        raise FileNotFoundError(f"No CSVs found in {FOLDER}")

    results = []
    for fp in files:
        try:
            out = process_file(fp)
            results.append(out)
            print(f"[OK] {fp.name}: {len(out)} rows")
        except Exception as e:
            print(f"[WARN] Skipping {fp.name}: {e}")

    if not results:
        raise RuntimeError("No files could be processed.")

    combined = pd.concat(results, ignore_index=True)
    out_path = OUTDIR / "aggregated_physical_crossborder_flows_2011.csv"
    combined.to_csv(out_path, index=False, encoding="utf-8")
    print(f"Saved: {out_path.resolve()}")

if __name__ == "__main__":
    main()
```

Load

```python
results = []

for i, country_code in enumerate(country_codes):
    print(f"{country_code}")

    try:
        param_load = client.query_load(country_code, start=start, end=end)

        results.append({
            "country_code": country_code,
            #"mean (MW)": param_load.mean(),
            "max (MW)": param_load.max()[0],
            #"mode (MW)": param_load.mode().max()
        })

    except Exception as e:
        print(f"Error in load for: {country_code}. Error message: {str(e)}")
        continue

    print(f"Completed data processing for: {country_code}")
    if i < len(country_codes) - 1: # Only sleep if not the last country
        time.sleep(2) 

load_df = pd.DataFrame(results)
```

```python
load_df.to_csv(r'saved_data/' + f"load_df_{year}.csv", sep='\t', encoding='utf-8')
```

Installed capacity 

```python
results = []

for i, country_code in enumerate(country_codes):
    print(f"{country_code}")
    try:
        param_capacity = client.query_installed_generation_capacity(country_code, start=start, end=end)
        row = param_capacity.iloc[0].to_dict()
        row["country_code"] = country_code
        results.append(row)
    except Exception as e:
        print(f"Error in capacity for: {country_code}. Error message: {str(e)}")
        continue

    print(f"Completed data processing for: {country_code}")
    if i < len(country_codes) - 1:
        time.sleep(2)

capacity_df = pd.DataFrame(results)
```

```python
capacity_df.to_csv(r'saved_data/' + f"capacity_df_{year}.csv", sep='\t', encoding='utf-8')
```
